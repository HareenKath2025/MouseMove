```python
import time
import threading
from pynput import mouse, keyboard

# Flag to control the loop
running = True

def move_mouse():
    controller = mouse.Controller()
    while running:
        # Get current position
        current_pos = controller.position
        # Move slightly right and down, then back (to jiggle without drifting too far)
        controller.position = (current_pos[0] + 10, current_pos[1] + 10)
        time.sleep(0.1)  # Small delay for smooth movement
        controller.position = (current_pos[0] - 10, current_pos[1] - 10)
        time.sleep(2)  # Wait 2 seconds before next move

def on_press(key):
    global running
    if key == keyboard.KeyCode.from_char('e'):
        running = False
        print("Stopping mouse movement...")
        return False  # Stop the listener

# Start mouse movement in a separate thread
mouse_thread = threading.Thread(target=move_mouse)
mouse_thread.start()

# Start keyboard listener
print("Mouse mover started. Press 'e' to stop.")
with keyboard.Listener(on_press=on_press) as listener:
    listener.join()

# Wait for mouse thread to finish
mouse_thread.join()
print("Stopped.")
```
