```python
import pyautogui
import time

# Safety settings
pyautogui.FAILSAFE = True   # Move mouse to top-left corner to abort
pyautogui.PAUSE = 0.1       # Small pause between actions

MOVE_PX = 50        # ~1.5 cm on a typical 96 dpi screen
INTERVAL = 60       # Seconds between each nudge cycle

print("Teams keep-alive started. Move mouse to top-left corner to stop.")

try:
    while True:
        x, y = pyautogui.position()

        pyautogui.moveRel(-MOVE_PX, 0, duration=0.5)   # Left
        time.sleep(0.3)
        pyautogui.moveRel(MOVE_PX, 0, duration=0.5)    # Right (back to origin)
        time.sleep(0.3)
        pyautogui.moveRel(0, -MOVE_PX, duration=0.5)   # Up
        time.sleep(0.3)
        pyautogui.moveRel(0, MOVE_PX, duration=0.5)    # Down (back to origin)
        time.sleep(0.3)

        # Return exactly to original position
        pyautogui.moveTo(x, y, duration=0.3)

        print(f"Nudge done at {time.strftime('%H:%M:%S')} — next in {INTERVAL}s")
        time.sleep(INTERVAL)

except KeyboardInterrupt:
    print("Stopped by user (Ctrl+C).")
```
